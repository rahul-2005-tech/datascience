```python
from abc import ABC, abstractmethod


class Beverage(ABC):
    def __init__(self, base, liquid):
        self._base = base
        self._liquid = liquid

    def base(self):
        return self._base

    def liquid(self):
        return self._liquid

    @abstractmethod
    def prepare(self):
        pass

    def show_message(self):
        print("Here is your order:")


class Espresso(Beverage):
    def prepare(self):
        return f"Espresso: {self.base()} (strong shot)"


class Americano(Beverage):
    def prepare(self):
        return f"Americano: {self.base()} + {self.liquid()}"


class Latte(Beverage):
    def __init__(self, base, liquid, milk):
        super().__init__(base, liquid)
        self._milk = milk

    def milk(self):
        return self._milk

    def prepare(self):
        return f"Latte: {self.base()} + {self.liquid()} + {self.milk()}"


class Cappuccino(Beverage):
    def __init__(self, base, liquid, milk, foam):
        super().__init__(base, liquid)
        self._milk = milk
        self._foam = foam

    def prepare(self):
        return f"Cappuccino: {self.base()} + {self.liquid()} + {self._milk} + {self._foam}"


class Mocha(Beverage):
    def __init__(self, base, liquid, chocolate):
        super().__init__(base, liquid)
        self._chocolate = chocolate

    def prepare(self):
        return f"Mocha: {self.base()} + {self.liquid()} + {self._chocolate}"


class Frappe(Beverage):
    def __init__(self, base, liquid, ice):
        super().__init__(base, liquid)
        self._ice = ice

    def prepare(self):
        return f"Frappe: {self.base()} + {self.liquid()} + {self._ice}"


print("\n☕ Welcome to Coffee Machine ☕")
print("1. Espresso")
print("2. Americano")
print("3. Latte")
print("4. Cappuccino")
print("5. Mocha")
print("6. Frappe")

menu = {
    1: Espresso("Coffee", "No Water"),
    2: Americano("Coffee", "Water"),
    3: Latte("Coffee", "Water", "Milk"),
    4: Cappuccino("Coffee", "Water", "Milk", "Foam"),
    5: Mocha("Coffee", "Water", "Chocolate"),
    6: Frappe("Coffee", "Water", "Ice")
}

choice = int(input("Select your coffee (1-6): "))

drink = menu.get(choice)

if drink is None:
    print("Invalid choice!")
else:
    drink.show_message()
    print(drink.prepare())
```
